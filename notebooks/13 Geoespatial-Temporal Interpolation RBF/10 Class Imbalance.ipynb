# Completes

```python
threshold = 0.80

# keep the columns with less than 90% NaN 
df2 = df.loc[:, df.isnull().mean() < threshold].copy()

vals2 = df2.isna().sum() / len(df2) * 100
vals2 = vals2.sort_values(ascending=True)

for val in vals2.index:
    print(f"{val}: {vals2[val]:.2f}%")

len(df2.columns)
```

```python
import matplotlib.pyplot as plt
vals2.sort_values().plot(kind="hist", bins=100)
plt.xlabel("NaN ratio")
plt.ylabel("Count of columns")
plt.title(f"Columns with less than {threshold*100:.0f}% NaN values")
plt.show()
```

```python
# for the "bad" "IBD_EQR_Status", know the NaN ratio
bads = df[df['IBD_EQR_Status'] == 'Bad']
badsclean = df2[df2['IBD_EQR_Status'] == 'Bad']
# Calculate NaN ratios per column (no double division)
nan_ratio = bads.isna().mean() / len(bads)
nan_ratio_clean = badsclean.isna().mean() / len(badsclean)

# Plot comparison
plt.figure(figsize=(10,5))
plt.hist(nan_ratio, bins=20, alpha=0.5, label="Original df")
plt.hist(nan_ratio_clean, bins=20, alpha=0.5, label="Cleaned df")
plt.xlabel("NaN ratio per column")
plt.ylabel("Count of columns")
plt.title("NaN distribution in 'Bad' class before/after cleaning")
plt.legend()
plt.show()
```

```python
from scipy.stats import ks_2samp

stat, p = ks_2samp(nan_ratio, nan_ratio_clean)
print(f"KS statistic={stat:.3f}, p={p:.3f}")
```

# SMOTE-NC

```python
# %pip install -U pandas numpy scikit-learn imbalanced-learn
import numpy as np, pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder
from sklearn.impute import SimpleImputer
from imblearn.over_sampling import SMOTENC

from cuml.neighbors import NearestNeighbors as KNN

# --- date → features ---
class DateFeaturizer:
    """Expand datetime cols into year, month, day-of-year, sin, cos."""
    def __init__(self, cols): self.cols = cols
    def fit(self, X, y=None): return self
    def transform(self, X):
        X = X.copy()
        out = []
        for c in self.cols:
            d = pd.to_datetime(X[c], errors="coerce")
            doy = d.dt.dayofyear.astype(float)
            ang = 2*np.pi*(doy/365.25)
            out.append(pd.DataFrame({
                f"{c}__year":  d.dt.year.astype(float),
                f"{c}__month": d.dt.month.astype(float),
                f"{c}__doy":   doy,
                f"{c}__sin":   np.sin(ang),
                f"{c}__cos":   np.cos(ang),
            }, index=X.index))
        return pd.concat(out, axis=1) if out else pd.DataFrame(index=X.index)

def smotenc_balance(df, target="IBD_EQR_Status",
                    ordered_cats=("poor","bad","moderate","good","high"),
                    date_cols=None, cat_cols=None, num_cols=None,
                    k=5, random_state=13):
    df = df.copy()

    # drop rows with missing target
    df = df[df[target].notna()].reset_index(drop=True)
    y = df[target].astype(str).str.strip().str.lower()

    # infer column groups if not passed
    if date_cols is None:
        date_cols = [c for c in df.columns if c!=target and ("date" in c.lower() or
                        pd.api.types.is_datetime64_any_dtype(df[c]))]
    if cat_cols is None:
        cat_cols  = [c for c in df.select_dtypes(include=["object","category"]).columns if c!=target and c not in date_cols]
    if num_cols is None:
        num_cols  = [c for c in df.columns if c not in [target]+cat_cols+date_cols and pd.api.types.is_numeric_dtype(df[c])]

    # preprocess: impute → encode
    num_pipe  = Pipeline([("imp", SimpleImputer(strategy="median"))])
    cat_pipe  = Pipeline([
        ("imp", SimpleImputer(strategy="most_frequent")),   # no NaNs left
        ("enc", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1)),
    ])
    date_pipe = Pipeline([
        ("fe",  DateFeaturizer(date_cols)),
        ("imp", SimpleImputer(strategy="median")),
    ])

    pre = ColumnTransformer([
        ("num",  num_pipe,  num_cols),
        ("date", date_pipe, date_cols),
        ("cat",  cat_pipe,  cat_cols),
    ], remainder="drop", verbose_feature_names_out=False)

    X = df[num_cols + date_cols + cat_cols]
    X_mat = pre.fit_transform(X)

    # categorical feature indices in transformed space
    n_num   = len(num_cols)
    n_date  = 5*len(date_cols)  # year, month, doy, sin, cos per date col
    cat_idx = np.arange(n_num + n_date, n_num + n_date + len(cat_cols))

    # robust k if minorities are tiny
    vc = pd.Series(y).value_counts()
    k_eff = max(1, min(k, int(vc.min())-1)) if len(vc) else 1
    nn = KNN(n_neighbors=max(1, k_eff))

    sm = SMOTENC(categorical_features=cat_idx, k_neighbors=nn,
                 random_state=random_state, sampling_strategy="auto")

    X_res, y_res = sm.fit_resample(X_mat, y.to_numpy())
    return (pd.DataFrame(X_res,
                         columns=list(pre.get_feature_names_out())),
            pd.Series(y_res, name=target))

# --- usage ---
# X_bal, y_bal = smotenc_balance(df,
#     target="IBD_EQR_Status",
#     date_cols=["Date_SamplingOperation"])  # set if known
# print(y_bal.value_counts())
```
