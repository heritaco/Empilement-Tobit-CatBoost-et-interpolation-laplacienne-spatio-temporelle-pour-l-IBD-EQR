```python
# Categorical IBD_EQR_Status over time with a fixed Red–Yellow–Green palette.
# Gray base points; points turn category color as dates pass.
# Requirements: pandas, numpy, matplotlib, pillow

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation, PillowWriter
from matplotlib.patches import Patch
from pathlib import Path

# Fixed 5-class R–Y–G palette (order matters)
_RYG_ORDER  = ["poor", "bad", "moderate", "good", "high"]
_RYG_COLORS = {
    "poor":     "#B30000",  # deep red
    "bad":      "#E34A33",  # red-orange
    "moderate": "#FDD835",  # yellow
    "good":     "#73C476",  # light green
    "high":     "#1A9850",  # deep green
}

def plot_status_over_time_ryg(
    df: pd.DataFrame,
    x_col: str = "Longitude_Lambert93",
    y_col: str = "Latitude_Lambert93",
    date_col: str = "Date_SamplingOperation",
    status_col: str = "IBD_EQR_Status",
    max_frames: int = 60,
):
    """
    Static PNG: all sites gray + colored overlay by IBD_EQR_Status (poor..high).
    Animated GIF: points progressively colored by category as date increases.

    Unknown/NaN statuses remain gray.
    """

    # create visualizations directory if not exists
    os.makedirs("visualizations", exist_ok=True)
    out_gif = Path("visualizations") / f"{status_col}_over_time.gif"
    out_png = Path("visualizations") / f"{status_col}_over_time_static.png"

    df = df.copy()
    df[date_col] = pd.to_datetime(df[date_col], errors="coerce")
    df = df.dropna(subset=[x_col, y_col, date_col])

    # Base gray layer. This data will remain gray regardless of date.
    X_all = df[x_col].to_numpy()
    Y_all = df[y_col].to_numpy()

    # Normalize status text
    status_norm = df[status_col].astype("string").str.lower().str.strip()

    # Map to indices [0..4]; unknown -> NaN
    cls_to_idx = {c: i for i, c in enumerate(_RYG_ORDER)}
    cls_idx = status_norm.map(cls_to_idx)

    has_class = cls_idx.notna().to_numpy()
    X = df.loc[has_class, x_col].to_numpy()
    Y = df.loc[has_class, y_col].to_numpy()
    K = cls_idx.loc[has_class].astype(int).to_numpy()
    D = df.loc[has_class, date_col].to_numpy()

    # Bounds with margin
    def extent_with_margin(x, y, m=0.02):
        xmin, xmax = np.nanmin(x), np.nanmax(x)
        ymin, ymax = np.nanmin(y), np.nanmax(y)
        dx, dy = xmax - xmin, ymax - ymin
        return (xmin - m*dx, xmax + m*dx, ymin - m*dy, ymax + m*dy)

    xmin, xmax, ymin, ymax = extent_with_margin(X_all, Y_all)

    # Colors per class in fixed order
    class_colors = [_RYG_COLORS[c] for c in _RYG_ORDER]

    # ---------- Static PNG ----------
    fig, ax = plt.subplots(figsize=(7.5, 7.5))
    ax.scatter(X_all, Y_all, s=10, color="0.7", alpha=0.7, linewidths=0)  # gray base
    handles = []
    for i, cls in enumerate(_RYG_ORDER):
        m = (K == i)
        if not np.any(m):
            continue
        ax.scatter(X[m], Y[m], s=12, color=class_colors[i], linewidths=0, label=cls)
        handles.append(Patch(facecolor=class_colors[i], edgecolor="none", label=cls))
    ax.set_xlabel(x_col); ax.set_ylabel(y_col)
    ax.set_title(f"{status_col} (R–Y–G) over all dates")
    ax.set_aspect("equal", adjustable="box")
    ax.set_xlim(xmin, xmax); ax.set_ylim(ymin, ymax)
    if handles:
        ax.legend(handles=handles, title=status_col, loc="best", frameon=True)
    fig.tight_layout()
    fig.savefig(out_png, dpi=150)
    plt.close(fig)

    # ---------- Animated GIF ----------
    dates = np.sort(df[date_col].unique())
    if len(dates) > max_frames:
        idx = np.linspace(0, len(dates)-1, max_frames).astype(int)
        dates = dates[idx]

    fig2, ax2 = plt.subplots(figsize=(7.5, 7.5))
    ax2.scatter(X_all, Y_all, s=10, color="0.7", alpha=0.7, linewidths=0)  # gray base

    scatters = []
    legend_handles = []
    for i, cls in enumerate(_RYG_ORDER):
        s = ax2.scatter([], [], s=12, color=class_colors[i], linewidths=0, label=cls)
        scatters.append(s)
        legend_handles.append(Patch(facecolor=class_colors[i], edgecolor="none", label=cls))

    ax2.set_xlabel(x_col); ax2.set_ylabel(y_col)
    ax2.set_aspect("equal", adjustable="box")
    ax2.set_xlim(xmin, xmax); ax2.set_ylim(ymin, ymax)
    ax2.legend(handles=legend_handles, title=status_col, loc="best", frameon=True)
    txt = ax2.text(0.02, 0.98, "", transform=ax2.transAxes, va="top", ha="left")

    def init():
        for s in scatters:
            s.set_offsets(np.empty((0, 2)))
        txt.set_text("")
        ax2.set_title(f"{status_col} (R–Y–G) over time")
        return (*scatters, txt)

    def update(i_frame):
        t = dates[i_frame]
        for k in range(len(_RYG_ORDER)):
            mask = (K == k) & (D <= t)
            offs = np.column_stack([X[mask], Y[mask]]) if np.any(mask) else np.empty((0, 2))
            scatters[k].set_offsets(offs)
        txt.set_text(f"Date ≤ {pd.to_datetime(t).date()}")
        return (*scatters, txt)

    anim = FuncAnimation(fig2, update, frames=len(dates), init_func=init, blit=True, interval=100)

    saved_gif = True
    try:
        anim.save(out_gif, writer=PillowWriter(fps=10))
    except Exception:
        saved_gif = False
    plt.close(fig2)

    return {"png": str(out_png), "gif": str(out_gif) if saved_gif else None, "frames": len(dates)}

# --- Usage ---
# results = plot_status_over_time_ryg(df)  # df must have Date_SamplingOperation, IBD_EQR_Status, and Lambert93 coords
# print(results)
```
