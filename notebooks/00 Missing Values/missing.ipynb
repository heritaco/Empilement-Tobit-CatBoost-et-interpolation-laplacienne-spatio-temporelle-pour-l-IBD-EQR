```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

def compare_missing_superposed(X, column, a, b, max_rows=800, order="diff"):
    """
    X: DataFrame
    column: str, the column used to filter into two groups
    a, b: values of `column` to compare
    max_rows: sample cap per group for heatmap speed
    order: "diff" -> sort columns by |pA - pB| desc,
           "max"  -> by max(pA, pB) desc,
           "name" -> alphabetical
    """
    XA = X[X[column] == a]
    XB = X[X[column] == b]
    if XA.empty or XB.empty:
        raise ValueError("One of the filtered subsets is empty.")

    # Per-column missing rates (use all rows for stable rates)
    pA = XA.isna().mean(numeric_only=False)
    pB = XB.isna().mean(numeric_only=False)

    # Choose a column order
    if order == "diff":
        sort_key = (pA - pB).abs().sort_values(ascending=False)
    elif order == "max":
        sort_key = pd.concat([pA, pB], axis=1).max(axis=1).sort_values(ascending=False)
    else:
        sort_key = pd.Series(index=sorted(X.columns), dtype=float)

    cols = sort_key.index.tolist() if len(sort_key) else list(X.columns)
    # Keep only columns present in both rate Series
    cols = [c for c in cols if c in pA.index and c in pB.index]

    # Sample rows (for heatmaps only) to keep render fast and comparable
    nA = min(len(XA), max_rows)
    nB = min(len(XB), max_rows)
    XA_s = XA.sample(n=nA, random_state=0) if len(XA) > nA else XA
    XB_s = XB.sample(n=nB, random_state=0) if len(XB) > nB else XB

    # Build figure: overlay rates + two synchronized heatmaps
    fig = plt.figure(figsize=(16, 10))
    gs = fig.add_gridspec(nrows=3, ncols=1, height_ratios=[1.2, 2, 2], hspace=0.25)

    # (1) Overlay per-column missing rates
    ax0 = fig.add_subplot(gs[0, 0])
    x = np.arange(len(cols))
    ax0.plot(x, pA[cols].values, label=f"{column}={a}", linewidth=1.8, alpha=0.3)
    ax0.plot(x, pB[cols].values, label=f"{column}={b}", linewidth=1.8, alpha=0.3)
    ax0.set_ylim(0, 1)
    ax0.set_ylabel("Missing rate")
    ax0.set_xticks([])
    ax0.legend(loc="upper right", frameon=False)
    ax0.set_title("Missingness comparison: overlayed per-column rates")

    # (2) Heatmap for group A
    ax1 = fig.add_subplot(gs[1, 0])
    sns.heatmap(XA_s[cols].isna(), cbar=False, yticklabels=False, ax=ax1)
    ax1.set_title(f"Row-level mask • {column}={a}")
    ax1.set_xticklabels([])
    

    # (3) Heatmap for group B (aligned columns)
    ax2 = fig.add_subplot(gs[2, 0], sharex=ax1)
    hm = sns.heatmap(XB_s[cols].isna(), cbar=False, yticklabels=False, ax=ax2)
    ax2.set_title(f"Row-level mask • {column}={b}")
    ax2.set_xticklabels([])

    plt.show()

def missing_values_compare(Column, var_a, var_b, max_rows=1000, order="max"):
    """
    Compare missing values between two subsets of Xtr defined by Column == var_a and Column == var_b.
    order has three options: 
    "diff" -> sort columns by |pA - pB| desc, where pA, pB are the missing rates in each subset,
    "max"  -> by max(pA, pB) desc,  
    "name" -> alphabetical
    """
    compare_missing_superposed(Xtr, Column, var_a, var_b, max_rows=max_rows, order=order)
```
