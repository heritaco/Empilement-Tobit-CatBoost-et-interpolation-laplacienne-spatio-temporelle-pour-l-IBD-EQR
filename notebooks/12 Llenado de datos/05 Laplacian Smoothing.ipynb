```python
# Laplacian smoothing for wide taxa table (harmonic extension).
# Keeps observed values; fills only NaNs.
# Requires: numpy, pandas, scipy, scikit-learn

import numpy as np
import pandas as pd
from sklearn.neighbors import NearestNeighbors
import scipy.sparse as sp
import scipy.sparse.linalg as spla

# ----------------------------
# Build kNN graph and Laplacian
# ----------------------------
def build_st_graph(meta: pd.DataFrame,
                   x_col="Longitude_Lambert93",
                   y_col="Latitude_Lambert93",
                   date_col="Date_SamplingOperation",
                   k=12, hs_km=80.0, ht_days=120.0,
                   symmetrize=True):
    """Returns (W, L, meta_out, Z) where L=D-W (unnormalized)."""
    m = meta.copy()
    m[date_col] = pd.to_datetime(m[date_col], errors="coerce")
    if m[date_col].isna().any():
        raise ValueError("Unparseable dates.")
    # Scale: meters→km; days unchanged
    hs_m = max(hs_km, 1e-6) * 1000.0
    t0 = m[date_col].min()
    Z = np.column_stack([
        m[x_col].to_numpy(float) / hs_m,
        m[y_col].to_numpy(float) / hs_m,
        (m[date_col] - t0).dt.days.to_numpy(float) / max(ht_days, 1e-6),
    ])
    n = len(m)
    k_eff = min(k, n-1) if n > 1 else 0
    if k_eff <= 0:
        W = sp.csr_matrix((n, n))
        D = sp.csr_matrix((n, n))
        L = D - W
        return W, L, m, Z

    nn = NearestNeighbors(n_neighbors=k_eff, metric="euclidean").fit(Z)
    d, idx = nn.kneighbors(Z, return_distance=True)  # (n, k)
    rows = np.repeat(np.arange(n), k_eff)
    cols = idx.ravel()
    w = np.exp(-0.5 * (d.ravel() ** 2))  # Gaussian weight on scaled distances
    W = sp.csr_matrix((w, (rows, cols)), shape=(n, n))
    if symmetrize:
        W = W.maximum(W.T)
    D = sp.diags(np.asarray(W.sum(axis=1)).ravel())
    L = D - W
    return W, L, m, Z

# ---------------------------------
# Harmonic extension (per taxon col)
# ---------------------------------
def laplacian_impute_matrix(M: pd.DataFrame,
                            L: sp.spmatrix,
                            obs_exact=True,
                            ridge=1e-6,
                            solver="spsolve"):
    """
    For each column y:
      Partition indices into O (observed) and M (missing).
      Solve (L_MM + ridge*I) z_M = - L_MO y_O, set z_O = y_O.
    Returns DataFrame same shape as M.
    """
    A = L.tocsr()
    n, p = M.shape
    Y = M.to_numpy(float)
    out = np.empty_like(Y)
    for j in range(p):
        y = Y[:, j]
        obs = np.isfinite(y)
        if obs.all():
            out[:, j] = y
            continue
        if not obs.any():
            out[:, j] = np.zeros(n)  # fallback; change if needed
            continue
        miss = ~obs
        # Sub-blocks
        L_MM = A[miss][:, miss]
        L_MO = A[miss][:, obs]
        b = - L_MO @ y[obs]
        # Stabilize
        A_sys = L_MM + ridge * sp.eye(L_MM.shape[0], format="csr")
        if solver == "cg":
            zM, info = spla.cg(A_sys, b, atol=0, rtol=1e-6, maxiter=2000)
            if info != 0:
                zM = spla.spsolve(A_sys.tocsc(), b)
        else:
            zM = spla.spsolve(A_sys.tocsc(), b)
        z = np.empty(n, dtype=float)
        z[obs]  = y[obs] if obs_exact else y[obs]  # clamp observed
        z[miss] = zM
        out[:, j] = z
    return pd.DataFrame(out, index=M.index, columns=M.columns)

# -------------------------------------------
# Wrapper for your df_long (wide taxa columns)
# -------------------------------------------
def impute_taxa_laplacian(df_long: pd.DataFrame,
                          id_col="SamplingOperations_code",
                          date_col="Date_SamplingOperation",
                          total_col="TotalAbundance_SamplingOperation",
                          x_col="Longitude_Lambert93",
                          y_col="Latitude_Lambert93",
                          taxa_cols=None,
                          k=12, hs_km=80.0, ht_days=120.0,
                          ridge=1e-6,
                          renorm_to_total=True):
    # Meta
    meta = (df_long[[id_col, date_col, total_col, x_col, y_col]]
            .drop_duplicates(subset=[id_col])
            .set_index(id_col)
            .sort_index())
    # Taxa matrix
    if taxa_cols is None:
        exclude = {id_col, date_col, total_col, x_col, y_col}
        taxa_cols = [c for c in df_long.columns
                     if c not in exclude and pd.api.types.is_numeric_dtype(df_long[c])]
    M = (df_long[[id_col] + taxa_cols]
         .groupby(id_col, as_index=True)
         .mean(numeric_only=True)
         .reindex(meta.index))

    # Graph
    _, L, meta_aligned, _ = build_st_graph(
        meta.reset_index(),
        x_col=x_col, y_col=y_col, date_col=date_col,
        k=k, hs_km=hs_km, ht_days=ht_days, symmetrize=True
    )
    L = L.tocsr()

    # Impute
    M_filled = laplacian_impute_matrix(M, L, obs_exact=True, ridge=ridge, solver="spsolve")

    # Optional: preserve per-sample totals by scaling only imputed part
    if renorm_to_total and total_col in meta_aligned.columns:
        totals = meta_aligned.set_index(id_col)[total_col].reindex(M_filled.index).to_numpy(float)
        obs_mask = np.isfinite(M.to_numpy(float))
        X = M.to_numpy(float)
        Y = M_filled.to_numpy(float)
        for i in range(Y.shape[0]):
            tgt = totals[i]
            if not np.isfinite(tgt):
                continue
            known_sum = np.nansum(np.where(obs_mask[i], X[i], 0.0))
            imp_mask = ~obs_mask[i]
            imp_sum = np.nansum(np.where(imp_mask, Y[i], 0.0))
            room = max(tgt - known_sum, 0.0)
            if imp_mask.any() and imp_sum > 0:
                Y[i, imp_mask] *= (room / imp_sum)
        M_filled = pd.DataFrame(Y, index=M_filled.index, columns=M_filled.columns)

    return M_filled, M, meta_aligned
```
