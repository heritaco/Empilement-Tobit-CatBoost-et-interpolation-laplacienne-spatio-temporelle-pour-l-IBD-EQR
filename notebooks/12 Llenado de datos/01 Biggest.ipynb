```python
import pandas as pd
import re

# 0) Harmonize names and types
em = envmeans.rename(columns={"Parametre_code":"Parameter_code",
                              "Parametre_label":"Parameter_label"}).copy()
em["Parameter_code"] = em["Parameter_code"].astype(str).str.strip()

ep = envparam.rename(columns={"Parameter_code":"Parameter_code"}).copy()
ep["Parameter_code"] = ep["Parameter_code"].astype(str).str.strip()
```

```python
# 1) Long form of thresholds with band labels
bands = [
    ("HighGood",       "Threshold_HighGood",       "Note_HighGood"),
    ("GoodModerate",   "Threshold_GoodModerate",   "Note_GoodModerate"),
    ("ModeratePoor",   "Threshold_ModeratePoor",   "Note_ModeratePoor"),
    ("PoorBad",        "Threshold_PoorBad",        "Note_PoorBad"),
]
parts = []
keep_meta = ["Parameter_code","Pressure","Parameter_label_eng",
             "Parameter_label_fr","Parameter_unit"]
for band, thr_col, note_col in bands:
    t = ep[keep_meta + [thr_col, note_col]].rename(
        columns={thr_col:"Threshold", note_col:"Note"}
    ).copy()
    t["Band"] = band
    parts.append(t)

ep_long = pd.concat(parts, ignore_index=True)
```

```python
# 2) Wide thresholds: one row per Parameter_code
def slug(x: str) -> str:
    x = "none" if pd.isna(x) or x=="" else str(x)
    return re.sub(r"\W+", "_", x.strip().lower())

ep_long["note_slug"] = ep_long["Note"].map(slug)
ep_long["colname"]   = "TH_" + ep_long["Band"] + "__" + ep_long["note_slug"]

thr_wide = (ep_long
            .pivot_table(index="Parameter_code",
                         columns="colname",
                         values="Threshold",
                         aggfunc="first"))

# choose one row of metadata per Parameter_code
meta = (ep_long
        .sort_values(["Parameter_code","Band"])  # stable pick
        .drop_duplicates("Parameter_code")
        [["Parameter_code","Pressure","Parameter_label_eng",
          "Parameter_label_fr","Parameter_unit"]]
        .set_index("Parameter_code"))

ep_wide = meta.join(thr_wide).reset_index()
```

```python






# 3) Merge with envmeans (keeps 1 row per OP×parameter; no duplication from ep_wide)
env_merged = em.merge(ep_wide, on="Parameter_code", how="left")

# Optionally sanity-check
# env_merged.sample(5)
# env_merged.columns  # now includes TH_HighGood__*, TH_GoodModerate__*, etc.
```

```python
import numpy as np
import pandas as pd

env_merged = env_merged.rename(columns={"Pressure_x":"Pressure",
                                        "Pressure_y":"Pressure_meta"})

# pick the first non-null per band by preference
prefer = ["none","maximum","minimum","ph_6_5","caco3_50_mg_l","caco3_200_mg_l","50_caco3_200_mg_l"]

for band in ["HighGood","GoodModerate","ModeratePoor","PoorBad"]:
    cols = [c for k in prefer
              for c in [f"TH_{band}__{k}"] if c in env_merged.columns]
    env_merged[f"TH_{band}"] = env_merged[cols].bfill(axis=1).iloc[:,0]

# keep tidy set
keep = ["CodeSite_SamplingOperations","SamplingOperations_code",
        "Parameter_code","Parameter_label","Pressure","Parameter_unit",
        "Mean1Y","Mean180Days","Mean90Days",
        "TH_HighGood","TH_GoodModerate","TH_ModeratePoor","TH_PoorBad"]
env_tidy = env_merged[keep].copy()
env_tidy
```

```python
import re
import pandas as pd

# env_tidy is your 12-col table shown in the message

def slug(s):
    s = str(s) if pd.notna(s) else ""
    return re.sub(r"\W+","_", s.strip().lower())

# stable, readable parameter key: "<code>_<label>"
env_tidy = env_tidy.copy()
env_tidy["param_key"] = (
    env_tidy["Parameter_code"].astype(str) + "_" + env_tidy["Parameter_label"].map(slug)
)

value_cols = ["Mean1Y","Mean180Days","Mean90Days",
              "TH_HighGood","TH_GoodModerate","TH_ModeratePoor","TH_PoorBad"]

```

```python

# pivot: one row per SamplingOperations_code; columns = param_key × value
wide = (env_tidy
        .pivot_table(index="SamplingOperations_code",
                     columns="param_key",
                     values=value_cols,
                     aggfunc="first")      # use "mean"/"max" if duplicates exist
       )

# flatten MultiIndex columns to "param__metric"
wide.columns = [f"{p}__{m}" for m,p in wide.columns]
wide = wide.sort_index(axis=1)
```

```python


# join to df (df can be indexed or not)
df_idx = df.set_index("SamplingOperations_code", drop=False)
final = df_idx.join(wide, how="left")
# final.index is SamplingOperations_code
```
