## Mapper

### Setup

```python
# Optional installs (uncomment if needed)
# %pip install kmapper umap-learn scikit-learn matplotlib pyvis --quiet

import numpy as np
import pandas as pd
from collections import Counter
from pathlib import Path

CLASS_ORDER = ['Bad','Poor','Moderate','Good','High']
PALETTE = {'Bad':'#b2182b','Poor':'#d6604d','Moderate':'#f1a340','Good':'#5aae61','High':'#1b7837'}
CLS2ID = {c:i for i,c in enumerate(CLASS_ORDER)}
```

### Helpers

```python
import kmapper as km
from sklearn.cluster import DBSCAN, AgglomerativeClustering
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from matplotlib.colors import ListedColormap

def _make_cover(n_cubes, overlap):
    for kw in (
        dict(n_cubes=n_cubes, perc_overlap=overlap),
        dict(n_cubes=n_cubes, overlap=overlap),
        dict(nr_cubes=n_cubes, perc_overlap=overlap),
        dict(nr_cubes=n_cubes, overlap=overlap),
    ):
        try: return km.Cover(**kw)
        except TypeError: pass
    raise TypeError("No compatible km.Cover(...) signature for this kmapper version")

def _mapper_map_robust(mapper, emb, Xs, cover, clusterer):
    try:
        return mapper.map(emb, X=Xs, cover=cover, clusterer=clusterer, remove_duplicate_nodes=True)
    except TypeError:
        return mapper.map(projection=emb, X=Xs, cover=cover, clusterer=clusterer, remove_duplicate_nodes=True)

def _visualize_robust(mapper, graph, y_pos, tooltips, out_html, title, cmap=None):
    trials = [
        dict(color_function=y_pos, custom_tooltips=tooltips, cmap=cmap),
        dict(color_function=y_pos, custom_tooltips=tooltips),
        dict(color_function=y_pos),
        dict(custom_tooltips=tooltips),
        dict()
    ]
    for kw in trials:
        try:
            return mapper.visualize(graph, path_html=out_html, title=title, **kw)
        except TypeError:
            continue
    return mapper.visualize(graph, path_html=out_html, title=title)

def _estimate_eps(Xs, k=10):
    from sklearn.neighbors import NearestNeighbors
    k = min(k+1, len(Xs))
    nn = NearestNeighbors(n_neighbors=k).fit(Xs)
    dists, _ = nn.kneighbors(Xs)
    kth = dists[:, -1]
    q = np.quantile(kth, [0.1, 0.2, 0.3, 0.5])
    return [float(v) for v in q]

def _degree_dict(links):
    deg = {}
    for a, nbrs in links.items():
        ai = int(a) if str(a).isdigit() else a
        deg.setdefault(ai, 0)
        for b in nbrs:
            bi = int(b) if str(b).isdigit() else b
            if ai == bi: 
                continue
            deg[ai] = deg.get(ai,0) + 1
            deg[bi] = deg.get(bi,0) + 1
    return deg
```

### Main

```python
def build_mapper_html_connected(
    X: pd.DataFrame,
    y: pd.Series,
    out_html_mapper: str = "mapper_ibd.html",
    out_html_majority: str = "mapper_ibd_majority.html",
    n_cubes: int = 8,           # fewer cubes → more connections
    overlap: float = 0.7,       # higher overlap → more edges
    random_state: int = 863
):
    # align
    if not X.index.equals(y.index):
        y = y.reindex(X.index)
    bad = set(pd.Series(y).dropna().unique()) - set(CLASS_ORDER)
    if bad:
        raise ValueError(f"Unknown labels: {bad}")

    # clean, impute, scale
    Xc = X.replace([np.inf, -np.inf], np.nan)
    Xi = SimpleImputer(strategy="median").fit_transform(Xc.values)
    Xs = StandardScaler().fit_transform(Xi)

    # lens: “stickier” UMAP
    try:

        import umap
        print("Using UMAP for lens...")
        emb = umap.UMAP(
            n_neighbors=30,
            min_dist=0.35,
            metric="euclidean",
            random_state=random_state
        ).fit_transform(Xs)
    except Exception:
        from sklearn.decomposition import PCA
        emb = PCA(n_components=2, random_state=random_state).fit_transform(Xs)

    mapper = km.KeplerMapper(verbose=0)

    # Try Agglomerative first (fewer isolates)
    cover = _make_cover(n_cubes=n_cubes, overlap=overlap)
    graph = None
    params_used = None
    for k in [3, 4, 5]:
        g = _mapper_map_robust(
            mapper, emb, Xs, cover,
            AgglomerativeClustering(n_clusters=k)
        )
        if len(g["nodes"]) > 0:
            graph = g
            params_used = dict(clusterer="Agglomerative", n_cubes=n_cubes, overlap=overlap, n_clusters=k)
            break

    # Fallback to DBSCAN with generous eps/min_samples
    if graph is None:
        base = _estimate_eps(Xs)[-1]  # median kNN radius
        for eps in [base*1.6, base*2.0, 1.6, 2.0, 2.5]:
            for ms in [2, 3]:
                g = _mapper_map_robust(
                    mapper, emb, Xs, cover,
                    DBSCAN(eps=float(eps), min_samples=int(ms), metric="euclidean")
                )
                if len(g["nodes"]) > 0:
                    graph = g
                    params_used = dict(clusterer="DBSCAN", n_cubes=n_cubes, overlap=overlap,
                                       eps=float(eps), min_samples=int(ms))
                    break
            if graph is not None:
                break

    if graph is None or len(graph["nodes"]) == 0:
        raise RuntimeError("No nodes. Try even fewer cubes (6) and overlap 0.75.")

    # color encoding to [0,1]
    y_ids = np.array([CLS2ID.get(lbl, np.nan) for lbl in y.values], dtype=float)
    if np.isnan(y_ids).any():
        y_ids[np.isnan(y_ids)] = (len(CLASS_ORDER)-1)/2.0
    y_pos = y_ids / (len(CLASS_ORDER)-1)

    # node diagnostics
    node_purity, node_entropy, node_majority = {}, {}, {}
    for node, members in graph["nodes"].items():
        labels = y.iloc[members]
        cnt = Counter(labels); N = len(members)
        probs = np.array([cnt.get(c,0)/N for c in CLASS_ORDER], float) if N else np.zeros(len(CLASS_ORDER))
        node_purity[node]  = probs.max() if N else np.nan
        node_entropy[node] = float(-np.sum(probs[probs>0]*np.log(probs[probs>0]))) if N else np.nan
        node_majority[node]= CLASS_ORDER[int(probs.argmax())] if N else None

    # somple_toolti
    cmap = ListedColormap([PALETTE[c] for c in CLASS_ORDER], name="IBDPalette")

    # HTML 1: KeplerMapper
    _visualize_robust(
        mapper, graph, y_pos=y_pos, tooltips=y,
        out_html=out_html_mapper, title="Mapper: IBD_EQR_Status", cmap=cmap
    )

    # HTML 2: PyVis majority color
    majority_html_written = None
    try:
        from pyvis.network import Network
        net = Network(height="800px", width="100%", notebook=False, directed=False)
        net.barnes_hut()
        for node, members in graph["nodes"].items():
            maj = node_majority[node]; pur = node_purity[node]; ent = node_entropy[node]
            title = f"Node {node}<br>n={len(members)}<br>majority={maj}<br>purity={pur:.3f}<br>entropy={ent:.3f}"
            net.add_node(int(node), label=str(node), color=PALETTE.get(maj, "#cccccc"), title=title)
        for a, nbrs in graph["links"].items():
            for b in nbrs:
                try: ai, bi = int(a), int(b)
                except: ai, bi = a, b
                if ai != bi:
                    net.add_edge(ai, bi)
        out_path = Path(out_html_majority)
        try: net.write_html(str(out_path))
        except AttributeError: net.show(str(out_path))
        majority_html_written = str(out_path.resolve())
    except Exception:
        majority_html_written = None

    # simple connectivity stats
    deg = _degree_dict(graph["links"])
    n_nodes = len(graph["nodes"])
    n_iso = sum(1 for v in deg.values() if v == 0)
    iso_ratio = n_iso / max(n_nodes, 1)

    return {
        "graph": graph,
        "params_used": params_used,
        "mapper_html": str(Path(out_html_mapper).resolve()),
        "majority_html": majority_html_written,
        "n_nodes": n_nodes,
        "isolated_nodes": n_iso,
        "isolate_ratio": iso_ratio,
        "node_purity": node_purity,
        "node_entropy": node_entropy,
        "node_majority": node_majority,
    }
```
