```python
taxon_cols = [c for c in df.columns if c not in {"IBD","TotalAbundance_SamplingOperation"}]
Xpm = df[taxon_cols].fillna(0.0).to_numpy(dtype=np.float32)     # per-mil
tot = df["TotalAbundance_SamplingOperation"].to_numpy(dtype=np.float32)

# counts and log-damped abundances
N = np.rint((tot[:,None] / 1000.0) * Xpm).astype(np.float32)    # n_ij
```

```python
Phi = np.log1p(N)                                               # φ_ij
Phi
```

```python
# ----- Option B: estimate proxy sensitivities s~ with v≡1 -----
b = df["IBD"].to_numpy(np.float32) * Phi.sum(axis=1)
b
```
