## What each field says

* **success: False, message: “The maximum number of iterations is exceeded.”**
  Reached `nit=100` without meeting stopping criteria.
* **status: 0**
  Iteration limit.
* **cost: 9.13e7**
  Objective ( \tfrac12|A x-b|_2^2 ). Large by itself is not informative; compare to baselines.
* **fun: [...]**
  Residual vector (r=A x-b) (some large ± values → mismatch on many rows).
* **optimality: 4.78e5**
  (|A^\top r|_\infty) at the projected point. Far from stationary → not solved yet.
* **active_mask: all 0**
  No variable is sitting exactly at a bound (your solution is interior (1<\tilde s_j<20)).
* **unbounded_sol: (..., rank=2180, singular values=...)**
  Unconstrained LSQ had components well outside ([1,20]) (e.g., 68.11). High collinearity; effective rank 2180 < 2292.


Interior solution consistent with bounds; higher (\tilde s_j) ≈ taxa associated with higher IBD in your data. Because the problem is ill-conditioned and under-converged, treat the exact values cautiously.

## Likely causes

1. **Ill-conditioning**: many sparse, correlated taxon columns; some taxa appear in very few rows.
2. **Scaling**: columns of wildly different norms.
3. **Tight iteration cap**: default `max_iter=100` is low for a 43 568×2292 system.

## Fix order (highest leverage first)

1. **Column filtering**: drop taxa with support < (k) rows (e.g., (k=30)).
2. **Column scaling**: scale each column to unit (\ell_2) norm, solve, then unscale:
   [
   A=A_0 D,\quad \min|A x-b| \iff \min|A_0 y-b|\ \text{with}\ y=D x.
   ]
3. **Increase work**: `max_iter=5000`, tighter `lsmr_tol=1e-8`.
4. **Method switch**: try `method="bvls"` for dense active-set box LSQ (can converge faster on box-constrained problems).
5. **Regularize** (optional, improves stability): ridge via augmentation
   [
   \min_x |A x-b|_2^2 + \lambda|x|_2^2
   \ \Longleftrightarrow
   \min_x \left|\begin{bmatrix}A\ \sqrt{\lambda}I\end{bmatrix}x-\begin{bmatrix}b\ 0\end{bmatrix}\right|_2^2.
   ]

````python
# -*- coding: utf-8 -*-
# Proxy-IBD sensitivities fit from wide per-mil matrix (sparse + bounded LS)
# Inputs:
#   df: pandas.DataFrame with columns:
#       - 'IBD'  (float in [1,20])
#       - 'TotalAbundance_SamplingOperation' (≈400)
#       - many taxon columns as per-mil abundances (0..1000, NaN for absent)
# Output:
#   s_tilde_df: DataFrame[taxon, s_tilde, at_lower, at_upper]
#   metrics printed: R2, MAE, RMSE, fraction-at-bounds

import numpy as np
import pandas as pd
import scipy.sparse as sp
from scipy.optimize import lsq_linear

# =========================
# CONFIG
# =========================
IBD_COL   = "IBD"
TOTAL_COL = "TotalAbundance_SamplingOperation"
META_COLS = {IBD_COL, TOTAL_COL, "SamplingOperations_code",
             "CodeSite_SamplingOperations", "Date_SamplingOperation"}

MIN_SUPPORT = 30      # drop taxa columns that are nonzero in < MIN_SUPPORT rows
USE_RIDGE   = True
RIDGE_LAMBDA = 1e-2   # small L2 regularization (0 to disable)
MAX_ITER    = 5000
LSMR_TOL    = 1e-8
TOL         = 1e-10

# =========================
# HELPERS
# =========================
def _coo_from_permil(df, taxon_cols, total_col):
    """Build COO of φ = log1p(n_ij) from per-mil abundances without densifying."""
    # Stack non-NaN per-mil entries
    pm = df[taxon_cols].stack(dropna=True)  # MultiIndex (row_label, col_name)
    if len(pm) == 0:
        raise ValueError("No non-NaN per-mil entries found.")

    # Optional: drop zeros if any (keep strictly positive per-mil)
    pm = pm[pm.values > 0]

    # Map row labels -> positions; col names -> positions
    row_pos = pd.Series(np.arange(len(df)), index=df.index)
    col_pos = pd.Series(np.arange(len(taxon_cols)), index=taxon_cols)

    rows = row_pos.loc[pm.index.get_level_values(0)].to_numpy()
    cols = col_pos.loc[pm.index.get_level_values(1)].to_numpy()
    vals_pm = pm.to_numpy(dtype=np.float64)

    # Counts n_ij = round((Total/1000)*pm_ij)
    totals = df[total_col]
    totals_sel = totals.loc[pm.index.get_level_values(0)].to_numpy(dtype=np.float64)
    n_ij = np.rint((totals_sel / 1000.0) * vals_pm)

    # φ_ij = log1p(n_ij). Keep strictly positive contributions.
    phi_vals = np.log1p(n_ij)
    keep = phi_vals > 0
    rows, cols, phi_vals = rows[keep], cols[keep], phi_vals[keep]

    shape = (len(df), len(taxon_cols))
    coo = sp.coo_matrix((phi_vals, (rows, cols)), shape=shape, dtype=np.float64)
    return coo.tocsr(), np.array(taxon_cols)

def _filter_cols_by_support(A, col_names, min_support):
    nz = A.getnnz(axis=0)
    keep = (nz >= min_support).A1 if sp.issparse(nz) else np.asarray(nz).ravel() >= min_support
    if keep.sum() == 0:
        raise ValueError("All columns dropped by MIN_SUPPORT; lower MIN_SUPPORT.")
    return A[:, keep], col_names[keep], keep

def _column_scale(A):
    # col_norm_j = ||A[:,j]||_2
    col_norm = np.sqrt(A.multiply(A).sum(axis=0)).A1
    col_norm[col_norm == 0] = 1.0
    inv = 1.0 / col_norm
    A_scaled = A @ sp.diags(inv, shape=(A.shape[1], A.shape[1]))
    return A_scaled, col_norm

def _augment_ridge(A, b, lam):
    # Min ||A x - b||^2 + λ||x||^2  <=>  Min ||[A; sqrt(λ)I] x - [b; 0]||^2
    n = A.shape[1]
    if lam <= 0:
        return A, b
    A_aug = sp.vstack([A, np.sqrt(lam) * sp.eye(n, dtype=np.float64)], format="csr")
    b_aug = np.concatenate([b, np.zeros(n, dtype=np.float64)])
    return A_aug, b_aug

# =========================
# MAIN FIT
# =========================
def fit_proxy_s_tilde(df):
    # 1) Identify taxon columns
    meta = [c for c in df.columns if c in META_COLS]
    candidate = [c for c in df.columns if c not in META_COLS]
    # keep only numeric candidates
    taxon_cols = [c for c in candidate if pd.api.types.is_numeric_dtype(df[c])]
    if IBD_COL not in df.columns or TOTAL_COL not in df.columns:
        raise KeyError("Missing required columns: 'IBD' and/or 'TotalAbundance_SamplingOperation'.")

    # 2) Build φ as CSR
    A0, taxon_cols = _coo_from_permil(df, taxon_cols, TOTAL_COL)

    # 3) Drop rows with zero sum φ
    row_sums = A0.sum(axis=1).A.ravel()
    mask_rows = row_sums > 0
    if mask_rows.sum() == 0:
        raise ValueError("All rows have zero φ. Check inputs.")
    A0 = A0[mask_rows]
    row_sums = row_sums[mask_rows]
    ibd = df.loc[mask_rows, IBD_COL].to_numpy(dtype=np.float64)

    # b = IBD * sum_j φ_ij
    b = ibd * row_sums

    # 4) Filter rare taxa
    A0, taxon_cols, keep_mask = _filter_cols_by_support(A0, taxon_cols, MIN_SUPPORT)

    # 5) Column-scale with correct bound scaling
    A, col_norm = _column_scale(A0)
    lower_y = 1.0 * col_norm
    upper_y = 20.0 * col_norm

    # 6) Optional ridge augmentation (stabilizes ill-conditioning)
    if USE_RIDGE:
        A_aug, b_aug = _augment_ridge(A, b, RIDGE_LAMBDA)
    else:
        A_aug, b_aug = A, b

    # 7) Solve bounded sparse LS (TRF+LSMR works with CSR)
    res = lsq_linear(
        A_aug, b_aug,
        bounds=(lower_y, upper_y),
        method="trf",
        lsq_solver="lsmr",
        lsmr_tol=LSMR_TOL,
        max_iter=MAX_ITER,
        tol=TOL,
        verbose=0,
    )
    y = res.x
    s_tilde = y / col_norm  # unscale back to s-space

    # 8) Evaluate on original (unscaled, filtered) system
    Nhat = A0 @ s_tilde                   # = sum_j φ_ij * s_tilde_j
    ibd_hat = Nhat / (row_sums + 1e-12)

    # Metrics
    y_true = ibd
    err = ibd_hat - y_true
    mae = float(np.mean(np.abs(err)))
    rmse = float(np.sqrt(np.mean(err**2)))
    r2 = float(1.0 - np.sum(err**2) / np.sum((y_true - y_true.mean())**2))

    at_lo = s_tilde <= (1.0 + 1e-6)
    at_hi = s_tilde >= (20.0 - 1e-6)
    frac_bounds = float(np.mean(at_lo | at_hi))

    print("=== Fit summary ===")
    print(f"status={res.status} success={res.success} nit={res.nit} message={res.message}")
    print(f"cost={res.cost:.6g} optimality={res.optimality:.6g}")
    print(f"Rows used={A0.shape[0]}  Taxa used={A0.shape[1]}")
    print(f"R2={r2:.4f}  MAE={mae:.4f}  RMSE={rmse:.4f}  frac_at_bounds={frac_bounds:.4f}")

    # 9) Package results
    s_tilde_df = pd.DataFrame({
        "taxon": taxon_cols,
        "s_tilde": s_tilde,
        "at_lower": at_lo,
        "at_upper": at_hi,
        "col_l2_norm": col_norm
    }).sort_values("s_tilde", ascending=False).reset_index(drop=True)

    # Also return row-level predictions aligned to original df (NaN where dropped)
    ibd_hat_full = np.full(len(df), np.nan, dtype=np.float64)
    ibd_hat_full[np.where(mask_rows)[0]] = ibd_hat

    return s_tilde_df, ibd_hat_full, res

# =========================
# USAGE
# =========================

````

````python
# Normalized proxy-IBD: W s ≈ IBD  (W rows sum to 1)
import numpy as np, pandas as pd, scipy.sparse as sp
from scipy.optimize import lsq_linear
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

IBD_COL, TOTAL_COL = "IBD", "TotalAbundance_SamplingOperation"
META = {IBD_COL, TOTAL_COL, "SamplingOperations_code", "CodeSite_SamplingOperations", "Date_SamplingOperation"}

LOWER, UPPER = 0.5, 22.0      # bounds for s
MIN_SUPPORT  = 1            # min nonzeros per taxon
RIDGE        = 3.0            # ridge λ for stability
MAX_ITER     = 8000
LSMR_TOL     = 1e-9
TOL          = 1e-12

def _build_phi(df, taxon_cols):
    stk = df[taxon_cols].stack(dropna=True)
    stk = stk[stk.values > 0]
    if stk.empty:
        raise ValueError("No positive per-mil entries.")
    row_idx = pd.Series(np.arange(len(df)), index=df.index)
    col_idx = pd.Series(np.arange(len(taxon_cols)), index=taxon_cols)
    r = row_idx.loc[stk.index.get_level_values(0)].to_numpy()
    c = col_idx.loc[stk.index.get_level_values(1)].to_numpy()
    pm = stk.to_numpy(dtype=np.float64)
    tot = df[TOTAL_COL].loc[stk.index.get_level_values(0)].to_numpy(dtype=np.float64)
    n = (tot / 1000.0) * pm
    phi = np.log1p(n)
    keep = phi > 0
    r, c, phi = r[keep], c[keep], phi[keep]
    return sp.csr_matrix((phi, (r, c)), shape=(len(df), len(taxon_cols)), dtype=np.float64)

def _row_normalize(A):
    rs = A.sum(axis=1).A.ravel()
    mask = rs > 0
    A = A[mask]
    rs = rs[mask]
    Dinv = sp.diags(1.0 / rs)
    W = Dinv @ A           # each row sums to 1
    return W, mask, rs

def _filter_support(A, cols, k):
    nz = A.getnnz(axis=0)
    keep = nz >= k
    return A[:, keep], cols[keep], keep

def _col_scale(A):
    cn = np.sqrt(A.multiply(A).sum(axis=0)).A1
    cn[cn == 0] = 1.0
    Asc = A @ sp.diags(1.0 / cn)
    return Asc, cn

def _augment_ridge(A, b, lam):
    if lam <= 0: return A, b
    n = A.shape[1]
    A_aug = sp.vstack([A, np.sqrt(lam) * sp.eye(n, dtype=np.float64)], format="csr")
    b_aug = np.concatenate([b, np.zeros(n, dtype=np.float64)])
    return A_aug, b_aug

def fit_proxy_ibd_normalized(df):
    # 0) columns
    taxon_cols = [c for c in df.columns if c not in META and pd.api.types.is_numeric_dtype(df[c])]
    assert IBD_COL in df and TOTAL_COL in df

    # 1) φ
    Phi = _build_phi(df, taxon_cols)

    # 2) W = row-normalized φ (drop zero-rows)
    W, mask_rows, rowsum_phi = _row_normalize(Phi)
    y = df.loc[mask_rows, IBD_COL].to_numpy(np.float64)  # target

    # 3) filter weak taxa on W
    Wf, taxon_cols_f, keep_cols = _filter_support(W, np.array(taxon_cols), MIN_SUPPORT)

    # 4) scale columns and scale bounds correctly
    A, col_norm = _col_scale(Wf)
    lower_y = LOWER * col_norm
    upper_y = UPPER * col_norm

    # 5) optional ridge
    A_aug, b_aug = _augment_ridge(A, y, RIDGE)

    # 6) bounded LS (TRF+LSMR supports CSR)
    res = lsq_linear(
        A_aug, b_aug,
        bounds=(lower_y, upper_y),
        method="trf", lsq_solver="lsmr",
        max_iter=MAX_ITER, lsmr_tol=LSMR_TOL, tol=TOL
    )
    ycoef = res.x
    s = ycoef / col_norm  # back to s-space

    # 7) predict on used rows/cols
    A_eval = W[:, keep_cols]
    ibd_hat = A_eval @ s

    # 8) metrics
    R2  = float(r2_score(y, ibd_hat))
    MAE = float(mean_absolute_error(y, ibd_hat))
    RMSE= float(np.sqrt(mean_squared_error(y, ibd_hat)))
    frac_bounds = float(np.mean((s <= LOWER+1e-6) | (s >= UPPER-1e-6)))

    print(f"status={res.status} success={res.success} nit={res.nit} {res.message}")
    print(f"Rows={A_eval.shape[0]} Taxa={A_eval.shape[1]}  R2={R2:.4f}  MAE={MAE:.4f}  RMSE={RMSE:.4f}  at_bounds={frac_bounds:.3f}")

    # align predictions to original df
    ibd_hat_full = np.full(len(df), np.nan, dtype=np.float64)
    ibd_hat_full[np.where(mask_rows)[0]] = np.asarray(ibd_hat).ravel()

    out = pd.DataFrame({
        "taxon": taxon_cols_f,
        "s_tilde": s,
        "at_lower": s <= LOWER+1e-6,
        "at_upper": s >= UPPER-1e-6,
        "col_l2_norm": col_norm
    }).sort_values("s_tilde", ascending=False).reset_index(drop=True)

    return out, ibd_hat_full, res

# Usage:
# s_tilde_df, ibd_hat_full, res = fit_proxy_ibd_normalized(df)
# df["IBD_hat_proxy"] = ibd_hat_full
# s_tilde_df.to_csv("s_tilde_taxa_normalized.csv", index=False)
````
