```python
import os
import polars as pl
import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

# Folder containing the CSV files
input_folder = "C:\\Users\\herie\\OneDrive - Fundacion Universidad de las Americas Puebla\\Universidad\\09 Semestre Otoño\\Clasificacion de Calidad del Agua IBD-EQR\\data\\Datathon_dataset\\Datathon_dataset"
output_folder = "C:\\Users\\herie\\OneDrive - Fundacion Universidad de las Americas Puebla\\Universidad\\09 Semestre Otoño\\Clasificacion de Calidad del Agua IBD-EQR\\data"

# Ensure the output folder exists
os.makedirs(output_folder, exist_ok=True)

# Iterate over all CSV files in the folder
for file_name in os.listdir(input_folder):
    print(" Procesando:" + file_name )
    if file_name.endswith(".csv"):
        csv_path = os.path.join(input_folder, file_name)
        parquet_path = os.path.join(output_folder, file_name.replace(".csv", ".parquet"))
        
        # Read CSV using Polars
        # Specify schema overrides or handle problematic values
        df = pl.read_csv(
            csv_path,
            schema_overrides={
                "CodeDepartement": pl.Utf8,  # Treat 'CodeDepartement' as string
                "Parametre_code": pl.Utf8   # Treat 'Parametre_code' as string to handle non-integer values like 'XOMP'
            },
        )

        
        # Convert to Parquet and save
        table = pa.Table.from_pandas(df.to_pandas())
        pq.write_table(table, parquet_path)
```
