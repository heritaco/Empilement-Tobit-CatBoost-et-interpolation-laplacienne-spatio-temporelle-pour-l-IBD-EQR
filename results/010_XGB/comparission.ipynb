```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns  
from scripts import style
style.mpl_apply()
```

```python
alldata = pd.read_csv('results/010_XGB/XGB_Truncated_bad_mapping.csv')
cleandata = pd.read_csv('results/010_XGB/XGB_Truncated_cleaned_bad_mapping.csv')
```

```python
alldata
```

```python
alldata = alldata.set_index('SamplingOperations_code')
cleandata = cleandata.set_index('SamplingOperations_code')
```

```python
# confusion matrix for all data vs cleaned data
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(alldata, cleandata)
```

```python
# plot the confusion matrix
labels = ['Bad', 'Poor', 'Moderate', 'Good', 'High']
plt.figure(figsize=(8, 6))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=labels, yticklabels=labels)
plt.xlabel('Model trained on Cleaned Data')
plt.ylabel('Model trained on All Data')
plt.title('Confusion Matrix Comparison')
plt.show()
```
